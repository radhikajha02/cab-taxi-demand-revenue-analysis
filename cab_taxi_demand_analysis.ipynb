```python
df["Pickup_DateTime"] = pd.to_datetime(df["Pickup_DateTime"])
df["Dropoff_DateTime"] = pd.to_datetime(df["Dropoff_DateTime"])
```

```python
df.dtypes
```

```python
# here we will create a new column This will allow us to answer questions like:
# What time of day has the highest cab demand?

df["Pickup_Hour"] = df["Pickup_DateTime"].dt.hour
```

```python
# here we will create a new column named pickup_day
# this will help us to look which days have the most cab trips
df["Pickup_Day"] = df["Pickup_DateTime"].dt.day_name()
```

```python
# from this we can analyze demand by month
df["Pickup_Month"] = df["Pickup_DateTime"].dt.month_name()
```

```python
# updated dataset
df.head()
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
# missing values
# for passenger count, here we will use the median
df["Passenger_Count"] = df["Passenger_Count"].fillna(
    df["Passenger_Count"].median()
)
# here i used the median because it is less affected by unusual values than the average.
```

```python
# for tip_amount i will replace missing values with 0
df["Tip_Amount"] = df["Tip_Amount"].fillna(0)
```

```python
# payment method
df["Payment_Method"] = df["Payment_Method"].fillna(
    df["Payment_Method"].mode()[0]
)
# here i will use mode , it will replace the missing values with the values which repeat the most
```

```python
# for driver rating i will use the median
df["Driver_Rating"] = df["Driver_Rating"].fillna(
    df["Driver_Rating"].median()
)
```

```python
df.isnull().sum()
```

```python
# now to check duplicate rows
df = df.isnull().sum()
```

```python
# now i will drop the duplicate records from the dataset
df = df.drop_duplicates()
```

```python
df.isnull().sum()
```

```python
# now we will check our new dataset
df.shape
```

```python
df = df[df["Fare_Amount"] > 0]
```

```python
# QUESTION 1
# "What time of day has the highest cab demand?"

# Count trips for each hour
hourly_trips = df["Pickup_Hour"].value_counts().sort_index()
```

```python
hourly_trips
```

```python
df["Pickup_Hour"].head(20)
```

```python
busiest_hour = hourly_trips.idxmax()
busiest_trips = hourly_trips.max()

print("Busiest hour:", busiest_hour)
print("Number of trips:", busiest_trips)
```

```python
# creating the hourly demand chart
plt.figure(figsize=(12, 5))

plt.plot(
    hourly_trips.index,
    hourly_trips.values,
    marker="o"
)

plt.title("Cab Demand by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Number of Trips")
plt.xticks(range(24))

plt.show()
```

```python
# least busy hour
least_busy_hour = hourly_trips.idxmin()
least_busy_trips = hourly_trips.min()

print("Least busy hour:", least_busy_hour)
print("Number of trips:", least_busy_trips)
```

```python
# Creating a time-period column
def get_time_period(hour):
    if 0 <= hour <= 5:
        return "Night"
    elif 6 <= hour <= 11:
        return "Morning"
    elif 12 <= hour <= 16:
        return "Afternoon"
    elif 17 <= hour <= 20:
        return "Evening"
    else:
        return "Late Night"


df["Time_Period"] = df["Pickup_Hour"].apply(get_time_period)
```

```python
df["Time_Period"].value_counts()
```

```python
time_period_trips = df["Time_Period"].value_counts()

time_period_trips
```

```python
plt.figure(figsize=(10, 6))

bars = plt.bar(
    time_period_trips.index,
    time_period_trips.values
)

plt.title("Cab Demand by Time Period", fontsize=16, fontweight="bold")
plt.xlabel("Time Period")
plt.ylabel("Number of Trips")

plt.xticks(rotation=0)

plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.show()
```

```python
# “Which cab type generates the most revenue per trip?”
# Calculate average fare per trip for each cab type

average_fare_by_cab = df.groupby("Cab_Type")["Fare_Amount"].mean()

average_fare_by_cab
```

```python
# Compare average fare across cab types

fig = px.bar(
    average_fare_by_cab,
    title="Average Fare Across Cab Types",
    color=average_fare_by_cab.index,
    color_discrete_sequence=["blue", "orange", "green", "red"]
)

fig.update_layout(
    xaxis_title="Cab Type",
    yaxis_title="Average Fare",
    showlegend=False
)

fig.show()
```

```python
# Which pickup locations should the company prioritize for cab availability?
# Calculate total revenue by pickup location

revenue_by_pickup = df.groupby("Pickup_Location")["Fare_Amount"].sum()

revenue_by_pickup.sort_values(ascending=False).head(10)
```

```python
# Visualize the top 10 pickup locations by revenue

top_revenue_pickup = revenue_by_pickup.sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 6))

colors = ["blue", "orange", "green", "red", "purple",
          "brown", "pink", "cyan", "gold", "gray"]

plt.barh(
    top_revenue_pickup.index[::-1],
    top_revenue_pickup.values[::-1],
    color=colors
)
plt.title("Top 10 Pickup Locations by Revenue", fontsize=16, fontweight="bold")
plt.xlabel("Total Revenue")
plt.ylabel("Pickup Location")
plt.show()
```
